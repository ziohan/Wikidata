Importing all the libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import ast
```

Reading the databases

```python
db = pd.read_csv('DBLP1_processed_.csv', sep = ';')
db.head()
```

Printing the number of lines and columns

```python
rows, columns = db.shape
print(f"Rows: {rows}, Columns: {columns}")
print("\n")
```

Printing all the types of data

```python
print(db.dtypes)
```

Printing the missing data as NaN and empty strings

```python
NaN = db.isna().sum()
empty = (db.applymap(lambda x: isinstance(x, str) and x.strip() == "")).sum()
print(f"NaN values: {NaN.sum()}")
print(f"Empty strings: {empty.sum()}")
print(f"Total missing values: {NaN.sum() + empty.sum()}")
```

Printing all the unique data by column (AJEITAR)

```python
for col in db.columns:
    print(f"Column name: {col}")
    
    values = db[col].dropna()
    flattened_values = []
    
    for v in values:
        if isinstance(v, str) and v.startswith('['):
            try:
                v = ast.literal_eval(v)
            except:
                pass
        
        if isinstance(v, (list, tuple, set)):
            flattened_values.extend(v)
        else:
            flattened_values.append(v)
    
    print(f"Total number of values: {len(flattened_values)}")
    print(f"Number of unique values: {len(set(flattened_values))}\n")
```

Printing cloned values (AJEITAR)

```python
import ast
from collections import Counter

def parse_value(v):
    if isinstance(v, str) and v.startswith('['):
        try:
            return ast.literal_eval(v)
        except:
            return v
    return v

for col in db.columns:
    print(f"\nColumn: {col}")
    
    values = db[col].dropna()
    flattened = []
    
    for v in values:
        v = parse_value(v)
        
        if isinstance(v, (list, tuple, set)):
            flattened.extend(v)
        else:
            flattened.append(v)
    
    counter = Counter(flattened)
    
    repeated = {k: v for k, v in counter.items() if v > 1}
    
    print(f"Repeated values ({len(repeated)}):")
    for k, v in repeated.items():
        print(f"{k} → {v} vezes")
```
